```python
import os
import cv2
import torch
import torchvision
from ultralytics import YOLO

# ==========================================================================
# TWO-STAGE DETECTION UNTUK VIDEO CCTV (JARAK JAUH)
# Tahap 1: Deteksi area orang (person) menggunakan YOLOv8n (COCO)
# Tahap 2: Crop area tubuh atas/tangan orang pada resolusi asli (HD),
#          lalu deteksi rokok menggunakan model hasil training (best.pt).
#          Cara ini mencegah tiang/objek ruangan terdeteksi sebagai rokok
#          dan membuat rokok berukuran kecil di CCTV terlihat jelas.
# ==========================================================================

person_model = YOLO('yolov8n.pt')                      # Model Tahap 1 (Deteksi Orang)
smoke_model = YOLO('runs/detect/train4/weights/best.pt') # Model Tahap 2 (Deteksi Rokok)

video_path = "D:/semester 5/PRAK PCD/Young_men_smoking_indoors_27ce285e.mp4"
output_dir = "runs/detect/predict_two_stage"
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, "Young_men_smoking_indoors_two_stage.mp4")

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
W = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
H = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
writer = cv2.VideoWriter(output_path, fourcc, fps, (W, H))

frame_idx = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frame_idx += 1

    # Tahap 1: Deteksi orang (class 0 = person)
    p_res = person_model(frame, classes=[0], conf=0.35, device=0, verbose=False)[0]
    p_boxes = p_res.boxes.xyxy.cpu().numpy().astype(int)

    all_cigs = []
    all_scores = []

    # Tahap 2: Crop area tiap orang (kepala s/d tangan) & deteksi batang rokok
    for (x1, y1, x2, y2) in p_boxes:
        pw, ph = x2 - x1, y2 - y1
        cx1 = max(0, x1 - int(pw * 0.12))
        cy1 = max(0, y1 - int(ph * 0.05))
        cx2 = min(W, x2 + int(pw * 0.12))
        cy2 = min(H, y1 + int(ph * 0.72))
        crop = frame[cy1:cy2, cx1:cx2]
        if crop.size == 0:
            continue

        s_res = smoke_model(crop, imgsz=640, conf=0.45, device=0, verbose=False)[0]
        for box in s_res.boxes:
            bx1, by1, bx2, by2 = box.xyxy[0].cpu().numpy()
            conf = float(box.conf[0].cpu().numpy())
            bw, bh = bx2 - bx1, by2 - by1

            # Filter ukuran: batang rokok tidak mungkin lebih besar dari 45% lebar / 35% tinggi badan orang
            if bw > pw * 0.45 or bh > ph * 0.35:
                continue

            # Kembalikan koordinat crop ke koordinat frame asli
            gx1, gy1, gx2, gy2 = cx1 + bx1, cy1 + by1, cx1 + bx2, cy1 + by2
            all_cigs.append([gx1, gy1, gx2, gy2])
            all_scores.append(conf)

    # Non-Maximum Suppression (NMS) global agar rokok di area orang yang berdekatan tidak dobel kotak
    if len(all_cigs) > 0:
        boxes_t = torch.tensor(all_cigs, dtype=torch.float32)
        scores_t = torch.tensor(all_scores, dtype=torch.float32)
        keep = torchvision.ops.nms(boxes_t, scores_t, iou_threshold=0.4)
        final_cigs = boxes_t[keep].numpy().astype(int)
        final_scores = scores_t[keep].numpy()
    else:
        final_cigs = []
        final_scores = []

    # Gambar bounding box pada frame
    for (gx1, gy1, gx2, gy2), conf in zip(final_cigs, final_scores):
        cv2.rectangle(frame, (gx1, gy1), (gx2, gy2), (0, 0, 255), 2)
        label = f"Rokok {conf:.2f}"
        (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.55, 2)
        cv2.rectangle(frame, (gx1, gy1 - th - 6), (gx1 + tw + 4, gy1), (0, 0, 255), -1)
        cv2.putText(frame, label, (gx1 + 2, gy1 - 4), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 2)

    writer.write(frame)
    if frame_idx % 30 == 0 or frame_idx == total_frames:
        print(f"Processed frame {frame_idx}/{total_frames} - Terdeteksi: {len(final_cigs)} rokok")

cap.release()
writer.release()
print(f"\nSelesai! Video hasil disimpan di: {output_path}")
```
